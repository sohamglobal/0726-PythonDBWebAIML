```python
a={1,2,3,4,5}
b={4,5,6,7,8}

# UNION - combine all values from both the sets
res=a.union(b)
print(res)

# INTERSECTION - common values from the sets
res=a.intersection(b)
print(res)

# DIFFERENCE - values in the first set but not in second
res=a.difference(b)
print(res)

# SYMMETRIC DIFFERENCE - values from both sets except the common
res=a.symmetric_difference(b)
print(res)
```

```python
amazon={'iphone','airpods','macbook','ipad','samsung tv','dell laptop'}
flipkart={'xbox','samsung tv','playstation','iphone','dell laptop','hp printer'}
print(amazon)
print(flipkart)
```

List all the products present on both shopping platforms

```python
all=amazon.union(flipkart)
print(all)
```

list of products available on both shopping platforms

```python
common=amazon.intersection(flipkart)
print(common)
```

show me a list of products available only on flipkart

```python
f=flipkart.difference(amazon)
print(f)
```

show me list of products only available on amazon

```python
a=amazon.difference(flipkart)
print(a)
```

show only the speciality items from both platforms

```python
s=amazon.symmetric_difference(flipkart)
print(s)
```
