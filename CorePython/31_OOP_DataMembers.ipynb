```python
class School:
  def show(self,scnm):
    self.scnm=scnm
    print(f"welcome to {self.scnm}")
  def display(self):
    print(f"thanks for joining {self.scnm}")
```

```python
obj=School()
```

```python
obj.show("soham international")
obj.display()
```

```python
class Person:
  name="praffull"
  def show(self):
    name="sharayu"
    print(name)
    print(self.name)
  def display(self):
    print(self.name)
```

```python
p=Person()
```

```python
p.show()
p.display()
```

```python
class Song:
  singer=""
  def store(self):
    singer="mohammed rafi"  #local variable of store()
    self.singer="michael jackson"
  def show(self):
    singer="kishore kumar"  #local variable of show()
    print(f"singer name {self.singer}")
    print(singer)
```

```python
s=Song()
s.store()
s.show()
```
