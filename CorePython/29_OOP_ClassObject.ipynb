Class -> is a collection of data and related functions

```python
class User:
  def welcomeuser(self):
    print('Hello user!')
    print('Welcome to OOP in Python')
    print('-'*35)
  def thanks(self,usernm):
    print(f'thanks {usernm} for joining sohamglobal')
```

```python
welcomeuser()
```

functions in the class can't run directly

to run functions in the class we have to create object of the class

```python
# User obj;   -> C++
# User obj=new User();  -> Java
obj=User()
```

```python
obj.welcomeuser()
```

every function in the class must have a parameter "self"

```python
obj.welcomeuser()
```

```python
class Calculator:
  def calcsquare(self,n):
    sq=n*n
    print(f"Square of {n} is {sq}")
  def addition(self,a,b):
    res=a+b
    print(f"sum of {a} and {b} is {res}")
```

```python
c=Calculator()
```

```python
c.addition(9,13)
```

```python
c.calcsquare(26)
```
