MAP - it will take all elements from the collection of values

```python
num=[1,2,3,4,5,6,7]
sq=list(map(lambda x:x**2,num))
print(sq)
```

```python
usd=[15,9,27,75,16,13]
rs=list(map(lambda x:x*95.73,usd))
print(rs)
```

FILTER - take only those elements which satisfy the condition

```python
num=[1,2,3,4,5,6,7,8,9]
odd=list(filter(lambda x:x%2==1,num))
print(odd)
```

```python
names=['microsoft','volkswagen','bmw','skoda','maruti','honda','oracle']
res=list(filter(lambda x:len(x)>8, names))
print(res)
```

REDUCE - it returns 1 result

```python
from functools import reduce
lst=[1,2,3,4,5,6,7,8,9]
sum=reduce(lambda x,y:x+y, lst)
print(sum)
```

```python
from functools import reduce
lst=[1,2,3,4,5,6,7,8,9]
res=reduce(lambda x,y:x*y, lst)
print(res)
```
