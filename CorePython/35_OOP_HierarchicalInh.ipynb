```python
class One:
  def calcsquare(self,n):
    sq=n**2
    print(f"square of {n} is {sq}")
```

```python
class Two(One):
  def addition(self,a,b):
    res=a+b
    print(f"sum of {a} and {b} is {res}")
```

```python
class Three(One):
  def finddiscount(self,amount):
    disc=amount*10/100
    print(f"discount will be {disc}")
```

```python
o1=Two()
o1.addition(44,66)
o1.calcsquare(77)
```

```python
o2=Three()
o2.finddiscount(6400)
o2.calcsquare(43)
```
