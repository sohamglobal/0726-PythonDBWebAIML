```python
import sys

lst=[1,2,3,4,5]
tup=(1,2,3,4,5)

print(sys.getsizeof(lst))
print(sys.getsizeof(tup))
```

```python
import timeit

list_time=timeit.timeit('x=[1,2,3,4,5]',number=100000)
tup_time=timeit.timeit('x=(1,2,3,4,5)',number=100000)

print(list_time)
print(tup_time)
```

LIST-> [], mutable, more memory, slow ||
TUPLE-> (), immutable, less memory, faster
