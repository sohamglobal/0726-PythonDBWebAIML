```python
from abc import ABC, abstractmethod
```

```python
class Payment(ABC):
  @abstractmethod   #decorator
  def pay(self):
    pass
```

```python
class UPI(Payment):
  def showsummary(self):
    print("summary of transactions...")
  def pay(self):
    print("UPI payment done...")
```

```python
obj=UPI()
obj.showsummary()
obj.pay()
```

```python
class CreditCard(Payment):
  def pay(self):
    print("credit payment processed successfully")
```

```python
cc=CreditCard()
cc.pay()
```
