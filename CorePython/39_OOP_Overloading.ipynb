Constructor overloading

```python
class Student:
    def __init__(self):
      print("constructor executed..object created in memory")
    def __init__(self,name):
      print(f"student name {name}")
    def show(self):
      print("welcome students to soham school")
    def __del__(self):
      print("object destroyed from the memory")
```

```python
obj=Student("sharayu")
obj.show()
```

```python
obj.show()
```

**Function overloading**

```python
class Calculator:
  def add(self,a,b):
    print(f"sum is {a+b}")
  def add(self,a,b,c):
    print(f"sum is {a+b+c}")
```

```python
c=Calculator()
c.add(23,78,33)
```

**Operator overloading**

```python
class Product:
  quantity=0
  purchcost=0
  def __init__(self,quantity,purchcost):
    self.quantity=quantity
    self.purchcost=purchcost
  def showstock(self):
    print(f"Quantity : {self.quantity}")
    print(f"Purchase cost : {self.purchcost}")
  def __add__(self,obj):
    q=self.quantity+obj.quantity
    c=self.purchcost+obj.purchcost
    return Product(q,c)
```

```python
dell=Product(13,945000)
dell.showstock()
```

```python
hp=Product(8,421000)
hp.showstock()
```

```python
total=dell+hp
total.showstock()
```
