```python
lst=[3,45,9,12,26,75,89,13]
sq=[]

for n in lst:
  sq.append(n**2)

print(sq)
```

```python
lst=[3,45,9,12,26,75,89,13]
sq=[n**2 for n in lst]
print(sq)
```

```python
nms=['praffull','sharayu','soham','shailaja','megha']
u=[nm.upper() for nm in nms]
print(u)
```

```python
lst=[9,13,1,26,45,30,10,15,18,25,79]
odd=[x for x in lst if x%2==1]
print(odd)
```

```python
lst=[9,13,1,26,-45,30,-10,15,-18,25,79]
neg=[n for n in lst if n<0]

print(neg)
```

```python
l1=[3,14,26,55,15]
l2=[1,7,13,2,9,26,14]
common=[x for x in l1 if x in l2]
print(common)
```

```python
l1=['java','python','sql','spring','spark','docker']
l2=['azure','java','oracle','cpp','spring']
res=[x for x in l1 if x not in l2]
res2=[x for x in l2 if x not in l1]
res.extend(res2)
print(res)
```

```python
print(len(res))
print(max(res))
print(min(res))
```
