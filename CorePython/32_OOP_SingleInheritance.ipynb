```python
# base class, parent class, super class
class One:
  def calcsquare(self,n):
    sq=n**2
    print(f"square of {n} is {sq}")
```

```python
# derived class, child class, sub class
class Two(One):
  def addition(self,a,b):
    res=a+b
    print(f"sum of {a} and {b} is {res}")
```

```python
obj=Two()
obj.addition(12,56)
obj.calcsquare(25)
```
