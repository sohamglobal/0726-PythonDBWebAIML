```python
def countzeros(lst):
  cnt=0
  for val in lst:
    if val==0:
      cnt+=1

  return cnt
```

```python
c=countzeros([9,26,1,0,13,10,0,0,45,34])
print(f"number of zeros in the list : {c}")
```

```python
c=countzeros([0,4,3,0,1,5,0,7,0,2,1,0,0,5])
print(c)
```
