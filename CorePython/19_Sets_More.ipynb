```python
n=[1,2,2,3,2,3,3,4,5,5,3,6,7,7,9]
print(n)
unique_list=list(set(n))
print(unique_list)
```

```python
allowed_users={'sharayu','praffull','shailaja'}
user=input('Enter userID : ')

if user in allowed_users:
  print('access granted')
else:
  print('access is denied')
```

```python
itemcodes=[1,2,3,2,1,2,3,4,2,5,1,2]
visited=set()

for item in itemcodes:
  if item not in visited:
    print(f"processing orders for {item}")
    visited.add(item)
```

**FrozenSet - Immutable**

```python
s={'java','python'}
print(s)
s.add('sql')
print(s)
```

```python
fs=frozenset([1,2,3,4,5])
print(fs)
```

```python
fs.add(7)
```

```python
l1=['java','spring boot','sql','jpa']
l2=['python','django','sql','flask']
```

```python
res=l1.intersection(l2)
print(res)
```

```python
res=list(set(l1).intersection(l2))
print(res)
```
