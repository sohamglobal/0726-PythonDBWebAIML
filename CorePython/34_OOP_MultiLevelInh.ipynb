```python
class One:
  def calcsquare(self,n):
    sq=n**2
    print(f"square of {n} is {sq}")
```

```python
class Two(One):
  def addition(self,a,b):
    res=a+b
    print(f"sum of {a} and {b} is {res}")
```

```python
class Three(Two):
  def finddiscount(self,amount):
    disc=amount*10/100
    print(f"discount will be {disc}")
```

```python
obj=Three()
obj.finddiscount(6310)
obj.calcsquare(9)
obj.addition(12,55)
```
