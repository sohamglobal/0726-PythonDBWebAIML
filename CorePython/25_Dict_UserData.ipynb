```python
dic={}
cho='yes'

while cho=='yes':
    prodid=int(input('Enter product ID : '))
    prodnm=input('Enter product name : ')
    dic[prodid]=prodnm
    cho=input('Do you want to continue (yes/no) : ')

print(dic)
```
